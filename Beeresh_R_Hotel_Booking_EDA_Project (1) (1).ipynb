## 5. Handle Missing Values

```python
missing_values = df.isnull().sum()

print(missing_values[missing_values > 0])
```

```python
plt.figure(figsize=(10,5))

missing = df.isnull().sum()
missing = missing[missing > 0].sort_values(ascending=False)

sns.barplot(x=missing.values, y=missing.index)

plt.title("Missing Values by Column")
plt.xlabel("Number of Missing Values")
plt.ylabel("Column")
plt.show()
```

```python
df.loc[:, "children"] = df["children"].fillna(0)
df.loc[:, "country"] = df["country"].fillna("Unknown")
df.loc[:, "agent"] = df["agent"].fillna(0)
df.loc[:, "company"] = df["company"].fillna(0)
```

```python
print("Missing values after handling:")
print(df[["children", "country", "agent", "company"]].isnull().sum())
```

```python
df.isnull().sum()
```

### Reason for handling missing values

Missing values in the children column are replaced with 0 because
a missing value can represent no recorded children.

Missing country values are replaced with "Unknown" because the
country information is unavailable.

Missing agent and company values are replaced with 0 to represent
the absence of a recorded agent or company.

## 9. Detect and Handle Outliers
### ADR Outlier Analysis

```python
plt.figure(figsize=(10,4))

sns.boxplot(x=df["adr"])

plt.title("ADR Outlier Analysis")
plt.xlabel("Average Daily Rate")

plt.show()
```

```python
Q1 = df["adr"].quantile(0.25)
Q3 = df["adr"].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

adr_outliers = df[
    (df["adr"] < lower_limit) |
    (df["adr"] > upper_limit)
]

print("Lower limit:", lower_limit)
print("Upper limit:", upper_limit)
print("Number of ADR outliers:", len(adr_outliers))
```

### Outlier Decision

The ADR boxplot shows potential outliers. These observations were
not automatically removed because unusually high hotel rates may
represent genuine premium or special bookings. Removing them without
business justification could remove useful information.

```python
month_order = [
    "January", "February", "March", "April",
    "May", "June", "July", "August",
    "September", "October", "November", "December"
]

monthly_bookings = (
    df["arrival_date_month"]
    .value_counts()
    .reindex(month_order)
)

print(monthly_bookings)
```

```python
plt.figure(figsize=(12,5))

monthly_bookings.plot(kind="bar")

plt.title("Bookings by Arrival Month")
plt.xlabel("Month")
plt.ylabel("Number of Bookings")
plt.xticks(rotation=45)

plt.show()
```
